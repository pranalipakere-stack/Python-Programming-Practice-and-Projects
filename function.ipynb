```python
def add(n1,n2):
    res=n1+n2
    return res
total=add(50,30)
print(total)
```

```python
def sub(n1,n2):
    res=n1-n2
    return res
total=sub(90,50)
print(total)
```

```python
def mul(n1,n2):
    res=n1*n2
    return res
total=mul(2,3)
print(total)
```

```python
def div(n1,n2):
    res=n1/n2
    return res
total=div(20,5)
print(total)
```

```python
def mod(n1,n2):
    res=n1%n2
    return res
total=mod(9,5)
print(total)
```

```python
#square
def sqr(n1):
    res=n1**2
    return res
total=sqr(3)
print(total)
```

```python
#square root
def sqr3(n1):
    res=n1**(1/2)
    return res
total=sqr3(25)
print(total)
```

```python
#area of circle 
def r(n1):
    area=3.14*n1*n1
    return area
total=r(2.3)
print(total)
```

```python
#area of square
def square(a):
    area=a*a
    return area
total=square(5)
print(total)
```

```python
#area of rectangle
def rect(l,w):
    area=l*w
    return area
total=rect(9,7)
print(total)
```

```python
def deposite(mbal,amt):
    mbal=mbal+amt
    return mbal
```

```python
def withdraw(mbal,amt):
    mbal=mbal-amt
    return mbal
```

```python
def balcheck(mbal):
    print(f"your current balance is : {mbal}")
```

```python
def mini(data):
    print("your last transaction :")
    for i in data:
        print(i)
```
